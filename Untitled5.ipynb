24.

```python
result = pd.pivot_table(
    df,
    values='units_sold',
    index='branch',
    columns='product',
    aggfunc='sum',
    fill_value=0,
    margins=True,
    margins_name='Total'
)

print(result)
```

25.

```python
result = pd.pivot_table(
    df,
    values=['discount_pct', 'customer_rating'],
    index='branch',
    columns='product',
    aggfunc='mean'
)

print(result)
```

26.

```python
print("Before Replacement:")
print(df[['branch', 'product', 'customer_rating']].to_string())

average_rating = df.groupby(
    ['branch', 'product']
)['customer_rating'].transform('mean')

df['customer_rating'] = df['customer_rating'].fillna(average_rating)

print("\nAfter Replacement:")
print(df[['branch', 'product', 'customer_rating']].to_string())
```

27.

```python
def weighted_avg(group):
    return (
        (group['units_sold'] * group['discount_pct']).sum()
        / group['units_sold'].sum()
    )

result = df.groupby('branch').apply(weighted_avg)

print(result)
```

28.

```python
result = df.groupby('branch', group_keys=False).apply(
    lambda x: x.nlargest(3, 'units_sold')
)

print(result[['branch', 'product', 'day', 'units_sold']])
```

29.

```python
branch_total = df.groupby('branch')['units_sold'].transform('sum')

df['sales_contribution_pct'] = (
    df['units_sold'] / branch_total
) * 100

print(df[['branch', 'units_sold', 'sales_contribution_pct']])

print("\nTotal Contribution of Each Branch:")
print(df.groupby('branch')['sales_contribution_pct'].sum())
```

30.

```python
df['sales_category'] = pd.qcut(
    df['units_sold'],
    q=4,
    labels=[
        'Low Sales',
        'Below Average Sales',
        'Above Average Sales',
        'High Sales'
    ]
)

result = df.groupby('sales_category', observed=False).agg({
    'units_sold': 'sum',
    'discount_pct': 'mean',
    'customer_rating': 'mean',
    'delivery_minutes': 'mean',
    'record_id': 'count'
})

result = result.rename(columns={
    'record_id': 'number_of_records'
})

print(result)
```
